```python
t = np.linspace(0, 10, 1000)
y_1 = np.sin(t)
y_2 = np.cos(t)
y_3 = y_1 + np.random.normal(0, 0.1, 1000)
data = np.column_stack((t, y_1, y_3))
np.savetxt("experiment_data.csv", data, delimiter = ",", fmt = "%.3f", header = "time,clean_signal,noisy_signal")
t_read, y_1_read, y_3_read = np.loadtxt("experiment_data.csv", delimiter = ",", skiprows = 1, unpack = True)
np.save('big_array.npy', np.column_stack((t_read, y_1_read, y_3_read)))
restored_arr = np.load('big_array.npy')
np.savez('lab_results.npz', time = t, sin = y_1, cos = y_2, noisy = y_3)
data_pack = np.load('lab_results.npz')
print(data_pack.files)
print(data_pack['cos'].shape)
```

```python
a = np.array([[2.0, 3.0], [4.0, -1.0]])
b = np.array([8.0, 2.0])
x = np.linalg.solve(a, b)
print(a @ x)
print(np.linalg.det(a))
a_inv = np.linalg.inv(a)
print(a_inv)
s = np.array([[2, 1],[3, 15]])
values, vectors = np.linalg.eig(s)
print(values)
print(vectors)
```

```python
a = np.array([[1, 2, 1],[3, -5, 2], [2, 7, -1]])
b = np.array([4, 1, 8])
x = np.linalg.solve(a, b)
print(np.linalg.norm(x))
```
